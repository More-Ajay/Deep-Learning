# ANN

```python
import numpy as np
import pandas as pd
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Sample real-life data: [Square Footage (scaled), Bedrooms, Age of House] , Ye hamara input/features hai , Har row ek house ko represent karti hai.
X_train = np.array([     # So, 3 input features hain , Square foot ko scaled form mein diya gaya hai(like 0.5 = 500 sqft).
    [0.5, 3, 10],
    [0.8, 4, 5],
    [0.3, 2, 20],
    [0.6, 3, 15]
])

# Corresponding house prices in thousands (e.g., 300 = $300,000) , Target / Output
y_train = np.array([300, 450, 200, 320])

# Build the ANN model
model = Sequential([                            # Sequential ka matlab: Neural network ki layers ko ek ke baad ek sequence mein arrange karna.
    Dense(6, input_dim=3, activation='relu'),  # Input layer + Hidden layer 1 with 6 neurons , Dense ek neural-network layer hai jisme har neuron previous layer ke har neuron se connected hota hai.
    Dense(6, activation='relu'),                 # Hidden layer 2 , input_dim=3 Hamare paas 3 input features hain , ReLU activation function - Neural network ko non-linear patterns learn karne mein help karta hai.
    Dense(1)                                     # Output layer for continuous value prediction , Dense(1) - Sirf 1 neuron kyun - Kyuki humein ek hi value predict karni hai , isme activation specify nahi kiya gaya , By default ye linear activation use karta hai , Regression problems ke liye ye common approach hai.
])

# Compile the model - Yahan hum model ko batate hain:Tum apni mistakes kaise measure karoge aur weights kaise improve karoge?
model.compile(optimizer='adam', loss='mean_squared_error') # Loss function - Loss batata hai model kitna galat predict kar raha hai , optimiser - Adam model ke weights ko adjust karta hai taaki loss kam ho.

# Train the model - Model ko training data se learn karao , X_train- input data , y_train-correct answer , Model input aur correct answer compare karke seekhta hai.
model.fit(X_train, y_train, epochs=50, verbose=0) # Epoch - Model ne poore training dataset ko ek baar process kiya , Model same training data ko 50 times learn karega. , Verbose=0 Iska matlab training ke time screen par progress print mat karo.

# Predict price for a new house: 700 sq ft (scaled as 0.7), 3 bedrooms, 8 years old
new_house = np.array([[0.7, 3, 8]]) # Ab hum ek new/unseen house model ko dete hain.
predicted_price = model.predict(new_house, verbose=0) # Ab trained ANN new house ko process karega , Exact prediction fixed nahi hai, kyunki neural network training random initialization ki wajah se different runs mein different result de sakta hai.

print(f"Predicted Price: ${predicted_price[0][0]:.2f} thousand")
```

```python
# 1. Create mock data for training [Credit Score, Income, Debt]
# Let's create a small history of 8 past applicants

X_train = np.array([
    [750, 8.0, 2.0],  # Good credit, high income, low debt -> Safe (1)
    [600, 3.5, 5.0],  # Poor credit, low income, high debt -> Default (0)
    [800, 12.0, 1.0], # Excellent credit, high income -> Safe (1)
    [550, 2.5, 4.0],  # Low credit -> Default (0)
    [710, 6.0, 3.0],  # Decent credit, moderate income -> Safe (1)
    [590, 4.0, 6.5],  # High debt -> Default (0)
    [680, 5.5, 2.0],  # Fair profile -> Safe (1)
    [620, 3.0, 3.5]   # Borderline profile -> Default (0)
], dtype=float)
```

```python
# Labels: 1 = Repaid loan, 0 = Defaulted
y_train = np.array([1, 0, 1, 0, 1, 0, 1, 0], dtype=float)
```

```python
# 2. Scale the data (Crucial step for Neural Networks to converge quickly)
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
```

```python
# 3. Build the Artificial Neural Network (ANN)

model = Sequential([
    Dense(units=4 , activation='relu' , input_shape=(3,)),
    Dense(units=1 , activation='sigmoid')
    ])
```

```python
# 4. Compile the model

model.compile(optimizer='adam' , loss='binary_crossentropy' , metrics=['accuracy'])
```

```python
# 5. Train the network

print('Training the ANN...')
model.fit(X_train_scaled , y_train , epochs=200 , verbose=0)
print('Training complete!\n')
```

```python
# 6. Test with a Brand New Applicant
# Profile: Credit score 730, Income $7.5k, Debt $1.5k (Looks like a safe bet!)

new_applicant = np.array([[730, 7.5, 1.5]])
new_applicant_scaled = scaler.transform(new_applicant)
```

```python
prediction = model.predict(new_applicant_scaled)
probability = prediction[0][0]

print("Approval Probability:",probability)
if probability >= 0.5:
    print("Decision: LOAN APPROVED ✅")
else:
    print("Decision: LOAN DENIED ❌")
```
